```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()

# Get the uploaded filename
filename = list(uploaded.keys())[0]

# Read the image
input_image = cv2.imread(filename)

# Convert to grayscale
input_image = cv2.cvtColor(input_image, cv2.COLOR_BGR2GRAY)

# Create 3x3 rectangular kernel
filterSize = (3, 3)
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, filterSize)

# Apply Black-Hat transformation
blackhat_img = cv2.morphologyEx(
    input_image,
    cv2.MORPH_BLACKHAT,
    kernel
)

# Display result
print("Black-Hat Image:")
cv2_imshow(blackhat_img)

# Save result
cv2.imwrite("blackhat.jpg", blackhat_img)

print("Image saved as blackhat.jpg")
```
