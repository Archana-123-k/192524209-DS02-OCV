```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read image
img = cv2.imread(filename)

if img is None:
    print("Error: Image could not be loaded.")
else:
    rows, cols = img.shape[:2]

    # Source points
    src_points = np.float32([
        [0, 0],
        [cols - 1, 0],
        [0, rows - 1],
        [cols - 1, rows - 1]
    ])

    # Destination points
    dst_points = np.float32([
        [50, 50],
        [cols - 50, 30],
        [30, rows - 50],
        [cols - 30, rows - 30]
    ])

    # Direct Linear Transformation
    # Calculate transformation matrix
    H, status = cv2.findHomography(src_points, dst_points, 0)

    # Apply transformation
    dlt_img = cv2.warpPerspective(img, H, (cols, rows))

    # Display original image
    print("Original Image:")
    cv2_imshow(img)

    # Display DLT transformed image
    print("DLT Transformed Image:")
    cv2_imshow(dlt_img)

    # Save output
    cv2.imwrite("DLT_Transformed_Image.jpg", dlt_img)

    print("Image saved as: DLT_Transformed_Image.jpg")
```
