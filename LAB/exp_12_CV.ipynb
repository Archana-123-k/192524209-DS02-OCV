```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload an image
uploaded = files.upload()

# Get uploaded filename
filename = list(uploaded.keys())[0]

# Read the image
frame = cv2.imread(filename)

# Check if image is loaded
if frame is None:
    print("Error: Image could not be loaded.")
else:
    # Define source ROI points
    roi_points = np.array([
        (150, 200),
        (450, 200),
        (550, 500),
        (50, 500)
    ])

    # Define target points
    target_points = np.array([
        (0, 0),
        (400, 0),
        (400, 600),
        (0, 600)
    ])

    # Convert points to float32
    roi_points = roi_points.astype(np.float32)
    target_points = target_points.astype(np.float32)

    # Calculate perspective transformation matrix
    M = cv2.getPerspectiveTransform(roi_points, target_points)

    # Apply perspective transformation
    dst = cv2.warpPerspective(frame, M, (400, 600))

    # Display original image
    print("Original Image:")
    cv2_imshow(frame)

    # Display transformed image
    print("Transformed Image:")
    cv2_imshow(dst)

    # Save the transformed image
    cv2.imwrite("Perspective_ROI_Transformed.jpg", dst)

    print("Image saved as: Perspective_ROI_Transformed.jpg")
```
