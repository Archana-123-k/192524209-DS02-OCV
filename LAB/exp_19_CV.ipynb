```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload the image
uploaded = files.upload()

# Get the uploaded filename
filename = list(uploaded.keys())[0]

# Read the image in grayscale
img = cv2.imread(filename, 0)

# Check if image is loaded
if img is None:
    print("Error: Image could not be loaded.")
else:
    # Apply Sobel operator in X direction
    sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)

    # Apply Sobel operator in Y direction
    sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

    # Combine Sobel X and Sobel Y
    edges = cv2.addWeighted(sobelx, 0.5, sobely, 0.5, 0)

    # Convert to 8-bit for proper display and saving
    edges = cv2.convertScaleAbs(edges)

    # Display original image
    print("Original Image:")
    cv2_imshow(img)

    # Display edge detected image
    print("Sobel Edge Detection:")
    cv2_imshow(edges)

    # Save the output
    cv2.imwrite("Edge_detection.jpg", edges)

    print("Image saved as: Edge_detection.jpg")
```
